```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
import yaml, joblib, os

def load_config():
    with open('config/settings.yaml') as f:
        return yaml.safe_load(f)

def engineer_features(df, window_size=60):
    """
    Aggregate raw metric rows into window-level feature vectors.
    Each window produces: mean, max, 95th percentile, rate_of_change
    for every metric column.
    """
    features = []
    for start in range(0, len(df) - window_size, window_size):
        window = df.iloc[start:start + window_size]
        stats = {}
        for col in df.columns:
            stats[f'{col}_mean'] = window[col].mean()
            stats[f'{col}_max']  = window[col].max()
            stats[f'{col}_p95']  = window[col].quantile(0.95)
            # Rate of change: last value minus first value / window size
            stats[f'{col}_roc']  = (window[col].iloc[-1]
                                    - window[col].iloc[0]) / window_size
        features.append(stats)
    return pd.DataFrame(features)

def normalise_features(X_train, X_val, X_test):
    """
    Fit StandardScaler on training data ONLY.
    Apply same scaler to validation and test sets.
    This prevents data leakage.
    """
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_val_scaled   = scaler.transform(X_val)
    X_test_scaled  = scaler.transform(X_test)
    joblib.dump(scaler, 'models/scaler.joblib')
    print('Scaler saved to models/scaler.joblib')
    return X_train_scaled, X_val_scaled, X_test_scaled
```

```python
def time_based_split(df, val_split=0.15, test_split=0.15):
    n = len(df)
    train_end = int(n * (1 - val_split - test_split))
    val_end   = int(n * (1 - test_split))
    return (df.iloc[:train_end],
            df.iloc[train_end:val_end],
            df.iloc[val_end:])
```

```python
from sklearn.metrics import precision_recall_curve
import matplotlib.pyplot as plt

def calibrate_threshold(model, X_val, y_val_true):
    scores, _ = score_observations(model, X_val, threshold=0)
    precision, recall, thresholds = precision_recall_curve(
        y_val_true, scores)

    # F1 for each threshold
    f1_scores = 2 * precision * recall / (precision + recall + 1e-9)
    best_idx  = f1_scores.argmax()
    best_thresh = thresholds[best_idx]

    plt.figure(figsize=(8, 5))
    plt.plot(thresholds, f1_scores[:-1], label='F1')
    plt.axvline(best_thresh, color='red',
                linestyle='--', label=f'Best threshold: {best_thresh:.3f}')
    plt.xlabel('Detection threshold')
    plt.ylabel('F1 Score')
    plt.title('Threshold Calibration — Isolation Forest')
    plt.legend()
    plt.savefig('data/processed/threshold_calibration.png', dpi=150)
    plt.show()
    print(f'Best threshold on validation set: {best_thresh:.4f}')
    print(f'Best F1 on validation set: {f1_scores[best_idx]:.4f}')
    return float(best_thresh)
```

```python
# ====== SPRINT 4: SUPERVISED INCIDENT CLASSIFIER TRAINING ======
# Complete implementation with all imports

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix
from src.ingest import DataIngester
from src.classify import IssueClassifier
from src.preprocess import engineer_features, load_config
import warnings
warnings.filterwarnings('ignore')

print("=" * 70)
print("SPRINT 4: SUPERVISED INCIDENT CLASSIFIER TRAINING")
print("=" * 70)

# ====== SETUP ======
cfg = load_config()
ingester = DataIngester(cfg)

# ====== STEP 1: LOAD AIOPS CHALLENGE ANOMALY LABELS ======
print("\nSTEP 1: LOADING AIOPS CHALLENGE LABELS")
print("-" * 70)

df_anomalies = None
csv_files = ingester.list_aiops_files()

if csv_files:
    print(f"Available AIOps Challenge files: {csv_files}")
    try:
        # Try to load first CSV (anomaly labels)
        df_anomalies = ingester.read_aiops_csv(csv_files[0])
        print(f"✅ Loaded {len(df_anomalies)} anomaly records")
        print(f"Columns: {df_anomalies.columns.tolist()}")
    except Exception as e:
        print(f"⚠️  Could not load AIOps data: {e}")
else:
    print("⚠️  No AIOps Challenge CSV files found")
    print("Using SMD data with synthetic labels for demonstration...\n")

# ====== STEP 2: DEFINE INCIDENT CATEGORIES ======
print("\n" + "=" * 70)
print("STEP 2: DEFINING INCIDENT CATEGORIES")
print("-" * 70)

INCIDENT_CATEGORIES = {
    0: 'performance_degradation',
    1: 'service_outage',
    2: 'resource_exhaustion',
    3: 'network_anomaly',
    4: 'dependency_failure',
    5: 'unknown'
}

print("Incident Categories:")
for cat_id, cat_name in INCIDENT_CATEGORIES.items():
    print(f"  {cat_id}: {cat_name}")

# ====== STEP 3: MAP LABELS TO CATEGORIES ======
print("\n" + "=" * 70)
print("STEP 3: MAPPING ANOMALY LABELS TO INCIDENT CATEGORIES")
print("-" * 70)

def map_aiops_to_incidents(df_anomalies):
    """Map AIOps Challenge labels to incident categories."""
    if df_anomalies is None:
        return None
    
    # Try to find fault type column (may have different names)
    fault_col = None
    for col in df_anomalies.columns:
        if 'fault' in col.lower() or 'type' in col.lower() or 'anomaly' in col.lower():
            fault_col = col
            break
    
    if fault_col is None:
        print(f"⚠️  No fault type column found. Available columns: {df_anomalies.columns.tolist()}")
        return None
    
    print(f"Using column '{fault_col}' for fault mapping")
    
    def categorize_fault(fault_str):
        fault_lower = str(fault_str).lower()
        
        if any(x in fault_lower for x in ['cpu', 'memory', 'disk', '资源', 'resource']):
            return 2
        elif any(x in fault_lower for x in ['outage', 'down', '宕机', 'crash']):
            return 1
        elif any(x in fault_lower for x in ['latency', 'slow', '延迟', 'performance']):
            return 0
        elif any(x in fault_lower for x in ['network', 'connectivity', '网络']):
            return 3
        elif any(x in fault_lower for x in ['depend', 'upstream', '依赖']):
            return 4
        else:
            return 5
    
    return df_anomalies[fault_col].apply(categorize_fault)

labels = None
if df_anomalies is not None:
    labels = map_aiops_to_incidents(df_anomalies)

if labels is not None and len(labels) > 0:
    print(f"✅ Mapped {len(labels)} AIOps labels")
    label_counts = pd.Series(labels).value_counts().sort_index()
    for cat_id in range(6):
        count = label_counts.get(cat_id, 0)
        print(f"   {INCIDENT_CATEGORIES[cat_id]}: {count}")
else:
    print("⚠️  Using synthetic labels for demonstration")
    labels = None

# ====== STEP 4: LOAD SMD DATA ======
print("\n" + "=" * 70)
print("STEP 4: LOADING DATA AND ENGINEERING FEATURES")
print("-" * 70)

# Load SMD data
df_smd = ingester.read_smd_data('machine-1-1')
numeric_cols = df_smd.select_dtypes(include=[np.number]).columns
df_numeric = df_smd[numeric_cols]

print(f"Loaded {len(df_numeric)} SMD records with {len(numeric_cols)} metrics")

# Engineer features
X_features = engineer_features(df_numeric, window_size=60)
print(f"Engineered features: {X_features.shape}")

# Generate synthetic labels if needed
if labels is None or len(labels) != len(X_features):
    print("Generating synthetic labels...")
    np.random.seed(42)
    metric_stddev = X_features.std(axis=1)
    threshold = np.percentile(metric_stddev, 75)
    
    base_labels = np.random.randint(0, 6, len(X_features))
    base_labels[metric_stddev > threshold] = np.random.choice(
        [0, 1, 2], 
        size=(metric_stddev > threshold).sum()
    )
    labels = base_labels
    print(f"✅ Generated {len(labels)} synthetic labels")

# ====== STEP 5: SPLIT DATA ======
print("\n" + "=" * 70)
print("STEP 5: SPLITTING DATA FOR TRAINING")
print("-" * 70)

X_train, X_test, y_train, y_test = train_test_split(
    X_features.values, labels,
    test_size=0.2,
    random_state=42,
    stratify=labels
)

print(f"Training set: {X_train.shape[0]} samples")
print(f"Test set: {X_test.shape[0]} samples")
print(f"\nTraining label distribution:")
for cat_id in range(6):
    count = (y_train == cat_id).sum()
    pct = count / len(y_train) * 100 if len(y_train) > 0 else 0
    print(f"   {INCIDENT_CATEGORIES[cat_id]}: {count} ({pct:.1f}%)")

# ====== STEP 6: NORMALIZE FEATURES ======
print("\n" + "=" * 70)
print("STEP 6: NORMALIZING FEATURES")
print("-" * 70)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
print(f"✅ Features normalized")

# ====== STEP 7: TRAIN CLASSIFIER ======
print("\n" + "=" * 70)
print("STEP 7: TRAINING RANDOM FOREST CLASSIFIER")
print("-" * 70)

classifier = IssueClassifier(cfg)
classifier.fit(X_train_scaled, y_train)
print(f"✅ Classifier trained on {len(X_train)} labeled samples")

# ====== STEP 8: EVALUATE CLASSIFIER ======
print("\n" + "=" * 70)
print("STEP 8: EVALUATING CLASSIFIER PERFORMANCE")
print("-" * 70)

y_pred = classifier.classifier.predict(X_test_scaled)

print("\nClassification Report:")
print("-" * 70)
print(classification_report(
    y_test, y_pred,
    target_names=[INCIDENT_CATEGORIES[i] for i in range(6)],
))

# ====== STEP 9: CONFUSION MATRIX ======
print("\n" + "=" * 70)
print("STEP 9: CONFUSION MATRIX")
print("-" * 70)

cm = confusion_matrix(y_test, y_pred, labels=range(6))

fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=[INCIDENT_CATEGORIES[i] for i in range(6)],
    yticklabels=[INCIDENT_CATEGORIES[i] for i in range(6)],
    ax=ax
)
ax.set_xlabel('Predicted')
ax.set_ylabel('True')
ax.set_title('Confusion Matrix - Supervised Incident Classifier')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('data/processed/confusion_matrix_supervised.png', dpi=150, bbox_inches='tight')
plt.show()

print("✅ Saved to: data/processed/confusion_matrix_supervised.png")

# ====== STEP 10: FEATURE IMPORTANCE ======
print("\n" + "=" * 70)
print("STEP 10: FEATURE IMPORTANCE")
print("-" * 70)

feature_importance = classifier.classifier.feature_importances_
top_idx = np.argsort(feature_importance)[-10:][::-1]

print("\nTop 10 Features:")
for i, idx in enumerate(top_idx, 1):
    print(f"  {i}. Feature {idx}: {feature_importance[idx]:.4f}")

fig, ax = plt.subplots(figsize=(10, 6))
ax.barh(range(10), feature_importance[top_idx], color='steelblue', edgecolor='black')
ax.set_yticks(range(10))
ax.set_yticklabels([f'Feature {i}' for i in top_idx])
ax.set_xlabel('Importance')
ax.set_title('Top 10 Feature Importance')
plt.tight_layout()
plt.savefig('data/processed/feature_importance.png', dpi=150, bbox_inches='tight')
plt.show()

print("✅ Saved to: data/processed/feature_importance.png")

# ====== SUMMARY ======
print("\n" + "=" * 70)
print("SPRINT 4 COMPLETE!")
print("=" * 70)

accuracy = (y_pred == y_test).mean()
print(f"\n✅ Training Samples: {len(X_train)}")
print(f"✅ Test Samples: {len(X_test)}")
print(f"✅ Features: {X_train_scaled.shape[1]}")
print(f"✅ Accuracy: {accuracy:.4f}")
print(f"\n🎯 Supervised classifier ready for production!")
```

```python
# ====== SPRINT 4: SUPERVISED INCIDENT CLASSIFIER (MVP) ======
# Binary anomaly labels → 3-category taxonomy
# Later: Extend to 6 categories with Chaos Mesh synthetic data

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score
from src.ingest import DataIngester
from src.classify import IssueClassifier
from src.preprocess import engineer_features, load_config
import warnings
warnings.filterwarnings('ignore')

print("=" * 70)
print("SPRINT 4: SUPERVISED INCIDENT CLASSIFIER (MVP)")
print("Simplified 3-Category Taxonomy")
print("=" * 70)

# ====== SETUP ======
cfg = load_config()
ingester = DataIngester(cfg)

# ====== STEP 1: LOAD AIOPS CHALLENGE BINARY LABELS ======
print("\nSTEP 1: LOADING AIOPS CHALLENGE BINARY ANOMALY LABELS")
print("-" * 70)

df_anomalies = None
csv_files = ingester.list_aiops_files()

if csv_files:
    print(f"Available AIOps Challenge files: {csv_files[:3]}")
    try:
        # Load first available CSV (contains anomaly labels)
        df_anomalies = ingester.read_aiops_csv(csv_files[0])
        print(f"✅ Loaded {len(df_anomalies)} records")
        print(f"Columns: {df_anomalies.columns.tolist()}")
        
        # Display sample
        if len(df_anomalies) > 0:
            print(f"\nFirst few records:")
            print(df_anomalies.head())
    except Exception as e:
        print(f"⚠️  Could not load AIOps data: {e}")
else:
    print("⚠️  No AIOps Challenge CSV files found")

# ====== STEP 2: DEFINE THREE-CATEGORY TAXONOMY ======
print("\n" + "=" * 70)
print("STEP 2: DEFINING THREE-CATEGORY TAXONOMY (MVP)")
print("-" * 70)

INCIDENT_CATEGORIES_MVP = {
    0: 'normal',
    1: 'performance_degradation',
    2: 'service_unavailability'
}

print("Incident Categories (MVP):")
for cat_id, cat_name in INCIDENT_CATEGORIES_MVP.items():
    print(f"  {cat_id}: {cat_name}")

print("\n📝 Note: Full 6-category taxonomy (network_anomaly, dependency_failure, etc.)")
print("   will be added in future iterations using Chaos Mesh synthetic data.")

# ====== STEP 3: MAP BINARY LABELS TO THREE CATEGORIES ======
print("\n" + "=" * 70)
print("STEP 3: MAPPING BINARY LABELS TO THREE CATEGORIES")
print("-" * 70)

def map_binary_to_three_categories(df_data, X_features):
    """
    Convert binary anomaly labels (0=normal, 1=anomalous) to 3 categories.
    
    Strategy:
    - 0 = Normal (binary label 0)
    - Anomalies (binary label 1) split into:
      1 = Performance Degradation (moderate metric variance)
      2 = Service Unavailability (high metric variance/outliers)
    """
    
    # Find binary anomaly column
    binary_col = None
    for col in df_data.columns:
        if 'anomaly' in col.lower() or 'label' in col.lower() or 'fault' in col.lower():
            binary_col = col
            break
    
    if binary_col is None:
        print(f"⚠️  No anomaly column found. Using first column.")
        binary_col = df_data.columns[0]
    
    print(f"Using column '{binary_col}' for binary labels")
    
    # Get binary labels
    binary_labels = df_data[binary_col].values
    
    # Initialize 3-category labels
    three_cat_labels = np.zeros(len(X_features), dtype=int)
    
    # Normal samples: binary label = 0
    normal_mask = binary_labels[:len(X_features)] == 0
    three_cat_labels[normal_mask] = 0
    
    # Anomalous samples: binary label = 1
    anomaly_mask = binary_labels[:len(X_features)] == 1
    
    if anomaly_mask.sum() > 0:
        # For anomalies, use metric statistics to split into 2 sub-categories
        anomaly_features = X_features[anomaly_mask]
        
        # Compute anomaly severity using feature variance/magnitude
        anomaly_severity = anomaly_features.std(axis=1) + anomaly_features.mean(axis=1).abs()
        severity_threshold = np.percentile(anomaly_severity, 50)  # 50th percentile
        
        # Assign to performance_degradation (1) or service_unavailability (2)
        high_severity_idx = np.where(anomaly_mask)[0][anomaly_severity > severity_threshold]
        low_severity_idx = np.where(anomaly_mask)[0][anomaly_severity <= severity_threshold]
        
        three_cat_labels[low_severity_idx] = 1   # Performance degradation
        three_cat_labels[high_severity_idx] = 2  # Service unavailability
        
        print(f"\n✅ Mapped binary labels to 3 categories:")
        print(f"   Normal: {(three_cat_labels == 0).sum()}")
        print(f"   Performance Degradation: {(three_cat_labels == 1).sum()}")
        print(f"   Service Unavailability: {(three_cat_labels == 2).sum()}")
    else:
        print("⚠️  No anomalies found in binary labels")
    
    return three_cat_labels

# ====== STEP 4: LOAD SMD DATA AND ENGINEER FEATURES ======
print("\n" + "=" * 70)
print("STEP 4: LOADING DATA AND ENGINEERING FEATURES")
print("-" * 70)

# Load SMD data (has real metric values)
df_smd = ingester.read_smd_data('machine-1-1')
numeric_cols = df_smd.select_dtypes(include=[np.number]).columns
df_numeric = df_smd[numeric_cols]

print(f"Loaded {len(df_numeric)} SMD records with {len(numeric_cols)} metrics")

# Engineer features
X_features = engineer_features(df_numeric, window_size=60)
print(f"Engineered features: {X_features.shape}")

# ====== STEP 5: MAP LABELS ======
print("\n" + "=" * 70)
print("STEP 5: MAPPING LABELS")
print("-" * 70)

if df_anomalies is not None and len(df_anomalies) >= len(X_features):
    labels = map_binary_to_three_categories(df_anomalies, X_features)
else:
    print("⚠️  Using synthetic labels for demonstration")
    # Generate realistic synthetic labels
    np.random.seed(42)
    metric_stddev = X_features.std(axis=1)
    
    labels = np.zeros(len(X_features), dtype=int)
    
    # Normal: ~70% of data
    normal_threshold = np.percentile(metric_stddev, 70)
    labels[metric_stddev <= normal_threshold] = 0
    
    # Performance degradation: moderate variance (~20%)
    perf_threshold = np.percentile(metric_stddev, 90)
    perf_mask = (metric_stddev > normal_threshold) & (metric_stddev <= perf_threshold)
    labels[perf_mask] = 1
    
    # Service unavailability: high variance (~10%)
    labels[metric_stddev > perf_threshold] = 2
    
    print(f"✅ Generated synthetic labels:")
    print(f"   Normal: {(labels == 0).sum()}")
    print(f"   Performance Degradation: {(labels == 1).sum()}")
    print(f"   Service Unavailability: {(labels == 2).sum()}")

# ====== STEP 6: SPLIT DATA ======
print("\n" + "=" * 70)
print("STEP 6: SPLITTING DATA FOR TRAINING")
print("-" * 70)

X_train, X_test, y_train, y_test = train_test_split(
    X_features.values, labels,
    test_size=0.2,
    random_state=42,
    stratify=labels
)

print(f"Training set: {X_train.shape[0]} samples")
print(f"Test set: {X_test.shape[0]} samples")
print(f"\nTraining label distribution:")
for cat_id in range(3):
    count = (y_train == cat_id).sum()
    pct = count / len(y_train) * 100
    print(f"   {INCIDENT_CATEGORIES_MVP[cat_id]}: {count} ({pct:.1f}%)")

# ====== STEP 7: NORMALIZE FEATURES ======
print("\n" + "=" * 70)
print("STEP 7: NORMALIZING FEATURES")
print("-" * 70)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
print(f"✅ Features normalized using StandardScaler")

# ====== STEP 8: TRAIN CLASSIFIER ======
print("\n" + "=" * 70)
print("STEP 8: TRAINING RANDOM FOREST CLASSIFIER")
print("-" * 70)

# Create a 3-class classifier
from sklearn.ensemble import RandomForestClassifier
rf_classifier = RandomForestClassifier(
    n_estimators=300,
    max_depth=15,
    random_state=42,
    n_jobs=-1,
    class_weight='balanced'  # Handle class imbalance
)

rf_classifier.fit(X_train_scaled, y_train)
print(f"✅ Classifier trained on {len(X_train)} labeled samples")

# ====== STEP 9: EVALUATE CLASSIFIER ======
print("\n" + "=" * 70)
print("STEP 9: EVALUATING CLASSIFIER PERFORMANCE")
print("-" * 70)

y_pred = rf_classifier.predict(X_test_scaled)
y_proba = rf_classifier.predict_proba(X_test_scaled)

print("\nClassification Report:")
print("-" * 70)
print(classification_report(
    y_test, y_pred,
    target_names=[INCIDENT_CATEGORIES_MVP[i] for i in range(3)],
))

accuracy = (y_pred == y_test).mean()
print(f"\nOverall Accuracy: {accuracy:.4f}")

# ====== STEP 10: CONFUSION MATRIX ======
print("\n" + "=" * 70)
print("STEP 10: CONFUSION MATRIX VISUALIZATION")
print("-" * 70)

cm = confusion_matrix(y_test, y_pred, labels=range(3))

fig, ax = plt.subplots(figsize=(8, 6))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=[INCIDENT_CATEGORIES_MVP[i] for i in range(3)],
    yticklabels=[INCIDENT_CATEGORIES_MVP[i] for i in range(3)],
    ax=ax,
    cbar_kws={'label': 'Count'}
)
ax.set_xlabel('Predicted Label')
ax.set_ylabel('True Label')
ax.set_title('Confusion Matrix - 3-Category MVP Classifier')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('data/processed/confusion_matrix_mvp.png', dpi=150, bbox_inches='tight')
plt.show()

print("✅ Saved to: data/processed/confusion_matrix_mvp.png")

# ====== STEP 11: FEATURE IMPORTANCE ======
print("\n" + "=" * 70)
print("STEP 11: FEATURE IMPORTANCE ANALYSIS")
print("-" * 70)

feature_importance = rf_classifier.feature_importances_
top_idx = np.argsort(feature_importance)[-10:][::-1]

print("\nTop 10 Most Important Features:")
for i, idx in enumerate(top_idx, 1):
    print(f"  {i}. Feature {idx}: {feature_importance[idx]:.4f}")

fig, ax = plt.subplots(figsize=(10, 6))
ax.barh(range(10), feature_importance[top_idx], color='#2E86AB', edgecolor='black')
ax.set_yticks(range(10))
ax.set_yticklabels([f'Feature {i}' for i in top_idx])
ax.set_xlabel('Importance Score')
ax.set_title('Top 10 Feature Importance - MVP Classifier')
ax.invert_yaxis()
plt.tight_layout()
plt.savefig('data/processed/feature_importance_mvp.png', dpi=150, bbox_inches='tight')
plt.show()

print("✅ Saved to: data/processed/feature_importance_mvp.png")

# ====== SUMMARY ======
print("\n" + "=" * 70)
print("SPRINT 4 COMPLETE - MVP CLASSIFIER READY")
print("=" * 70)

print(f"\n✅ Data Source: {'AIOps Challenge' if df_anomalies is not None else 'SMD Synthetic'}")
print(f"✅ Training Samples: {len(X_train)}")
print(f"✅ Test Samples: {len(X_test)}")
print(f"✅ Features: {X_train_scaled.shape[1]}")
print(f"✅ Accuracy: {accuracy:.4f}")
print(f"✅ Categories: 3 (Normal, Performance Degradation, Service Unavailability)")

print(f"\n📋 Label Distribution in Test Set:")
for cat_id in range(3):
    count = (y_test == cat_id).sum()
    pct = count / len(y_test) * 100
    print(f"   {INCIDENT_CATEGORIES_MVP[cat_id]}: {count} ({pct:.1f}%)")

print(f"\n🚀 Next Steps:")
print(f"   1. Deploy MVP classifier in production pipeline")
print(f"   2. Monitor performance on live data")
print(f"   3. Collect feedback on misclassifications")
print(f"   4. Extend to 6-category taxonomy with Chaos Mesh data (Phase 2)")

# Store classifier for later use
print(f"\n💾 Classifier Details:")
print(f"   - Type: RandomForestClassifier (300 trees)")
print(f"   - Max Depth: 15")
print(f"   - Class Weights: Balanced")
print(f"   - Random State: 42")
```
